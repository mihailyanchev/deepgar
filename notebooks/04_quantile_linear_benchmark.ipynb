```python
def denormalize_sigma(x, scaler):
    if scaler is None:
        result = x
    else:
        # Same units as denormalize(): percent. The original omitted the *100,
        # making the predictive spread 100x too narrow (see README).
        result = x*scaler['std'].iloc[0]*100
    return result
```

## Non-simultaneous Quantile Estimation

```python
from tf_keras import backend as K

early_stop = keras.callbacks.EarlyStopping(monitor='val_loss', patience=20)

def tilted_loss(q,y,f):
    e = (y-f)
    return K.mean(K.maximum(q*e, (q-1)*e), axis=-1)

def keras_pred(early_stop, quantile):
    # Set input_dim for the number of features.
    if len(data['X_TRAIN'].shape) == 1:
        input_dim = 1
    else:
        input_dim = data['X_TRAIN'].shape[0]
    model = keras.Sequential([
      keras.layers.Dense(1, activation='linear')
    ])
    
    model.compile(loss=lambda y,f: tilted_loss(quantile,y,f), optimizer=keras.optimizers.Adam(0.001))
    model.fit(data['X_TRAIN'], data['Y_TRAIN'], epochs=200, batch_size=1,
              verbose=0, validation_split=0.2, callbacks=[early_stop])
    
    return model.predict(data['X_TEST']), model.predict(data['X_TRAIN']), model

forecast_test = np.empty([data['X_TEST'].shape[0],4])
counter = 0

for quantile in param['QUANTILES']:
    quantile_test, quantile_train, q_model = keras_pred(early_stop, quantile)
    # Predict the quantile
    print(quantile_test.shape)
    forecast_test[:,counter] = quantile_test.squeeze()
    counter += 1
    
```
